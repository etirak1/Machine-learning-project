Kolona TotalCharges  
8% vrijednosti dodato na početku i na kraju _!_  
od 5% vrijednosti je oduzet broj 5  
3% vrijednosti obrisana . kod decimalnog broja  
Kod 3% vrijednosti je dodata 0 na početku

```python
n8_tc = int(len(df) * 0.08)
idx_8_tc = df.sample(n8_tc, random_state=42).index

df.loc[idx_8_tc, 'TotalCharges'] = df.loc[idx_8_tc, 'TotalCharges'].astype(str).apply(lambda x: f"_!_{x}_!_")
```

```python
import numpy as np

numeric_mask = df['TotalCharges'].apply(lambda x: isinstance(x, (int, float, np.integer, np.floating)))

n5_numeric = int(numeric_mask.sum() * 0.05)
idx_5_numeric = df[numeric_mask].sample(n5_numeric, random_state=42).index

df.loc[idx_5_numeric, 'TotalCharges'] = df.loc[idx_5_numeric, 'TotalCharges'] - 5
```

```python
import numpy as np

# jer neke si već pretvorila u string sa _!_ i sl.
numeric_mask_tc = df['TotalCharges'].apply(lambda x: isinstance(x, (int, float, np.integer, np.floating)))

n3_tc = int(numeric_mask_tc.sum() * 0.03)
idx_3_tc = df[numeric_mask_tc].sample(n3_tc, random_state=42).index

df.loc[idx_3_tc, 'TotalCharges'] = df.loc[idx_3_tc, 'TotalCharges'].astype(str).str.replace('.', '', regex=False)
```

```python
n5_tc_prefix = int(len(df) * 0.03)
idx_5_tc_prefix = df.sample(n5_tc_prefix, random_state=42).index

df.loc[idx_5_tc_prefix, 'TotalCharges'] = \
    df.loc[idx_5_tc_prefix, 'TotalCharges'].astype(str).apply(lambda x: "0" + x)
```
